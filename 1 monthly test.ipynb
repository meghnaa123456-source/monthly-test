```python



def add():
    ID=int(input("enter ID:"))
    if ID in student:
        print("ID already exists")
    else:
        name=input("enter name:")
        marks=input("enter marks:")   
        student[ID]={"name":name,"mark":marks}
        print("student added successfully")
def view():
    if not student: 
        print("NO record ")
    else:
        print(student)
def search():
    ID= int(input("enter ID:"))
    print(student.get(ID,"Not found"))
def update():
    ID = int(input("enter ID:"))
    if ID in student:
        student[ID][1]=int(input("enter marks:"))
        print("marks update")
    else:
        print("not found")
                    
print("1.add student",)
print("2.view all student")
print("3.search student")
print("4.update")
print("5.exist")
                    
choice = input("enter choice:")
if choice=="1":
    add()
elif choice=="2":
    view()
elif choice=="3":
    search()    
elif choice=="4":
    update()
elif choice=="5":
    print("thank you")
else:
    print("invalid choice")
                    
```
